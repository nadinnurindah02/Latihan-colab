```python
# Program Menghitung Luas Lingkaran

import math

print("PROGRAM LUAS LINGKARAN")

# Input jari-jari
r = float(input("Masukkan jari-jari lingkaran: "))

# Rumus luas
luas = math.pi * r * r

# Output hasil
print("Luas lingkaran adalah =", luas)
```
