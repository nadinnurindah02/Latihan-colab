```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Penjelasan kode

NumPy digunakan untuk membantu perhitungan numerik.

Pandas digunakan untuk membuat tabel iterasi agar hasil perhitungan lebih mudah dibaca.

Matplotlib digunakan untuk menampilkan grafik fungsi.

```python
fungsi = "x**3 - 2*x - 5"

batas_bawah = 2
batas_atas = 3

toleransi = 0.0001
```

## Penjelasan Data

Pada bagian ini ditentukan fungsi yang akan dicari akarnya.

Batas bawah dan batas atas digunakan sebagai interval awal pencarian akar.

Toleransi digunakan sebagai batas ketelitian hasil perhitungan. Semakin kecil nilai toleransi, semakin teliti hasil yang diperoleh.

```python
def hitung_fungsi(x):
    return eval(fungsi)
```

## Penjelasan Fungsi

Fungsi ini digunakan untuk menghitung nilai fungsi yang dimasukkan pengguna.

Sebagai contoh, jika fungsi yang dimasukkan adalah x**3 - 2*x - 5, maka program dapat menghitung nilai fungsi tersebut pada setiap nilai x yang diperlukan.

```python
jumlah_iterasi = 0

while abs(batas_atas - batas_bawah) > toleransi:

    titik_tengah = (batas_bawah + batas_atas) / 2

    if hitung_fungsi(batas_bawah) * hitung_fungsi(titik_tengah) < 0:
        batas_atas = titik_tengah
    else:
        batas_bawah = titik_tengah

    jumlah_iterasi += 1

akar = (batas_bawah + batas_atas) / 2

print("Akar persamaan =", akar)
print("Jumlah iterasi =", jumlah_iterasi)
```

## Metode Bisection

Metode Bisection bekerja dengan membagi interval menjadi dua bagian yang sama besar.

Nilai tengah interval dihitung, kemudian ditentukan interval mana yang mengandung akar.

Proses ini dilakukan berulang kali sampai panjang interval lebih kecil dari nilai toleransi yang ditentukan.

## Membuat Tabel Iterasi Metode Bisection

Pada bagian ini ditampilkan proses pencarian akar pada setiap iterasi.

Tabel ini berisi nilai batas bawah, batas atas, titik tengah, dan nilai fungsi pada titik tengah.

```python
data_iterasi = []

batas_bawah = 2
batas_atas = 3
toleransi = 0.0001

iterasi = 0

while abs(batas_atas - batas_bawah) > toleransi:

    titik_tengah = (batas_bawah + batas_atas) / 2

    data_iterasi.append([
        iterasi + 1,
        batas_bawah,
        batas_atas,
        titik_tengah,
        hitung_fungsi(titik_tengah)
    ])

    if hitung_fungsi(batas_bawah) * hitung_fungsi(titik_tengah) < 0:
        batas_atas = titik_tengah
    else:
        batas_bawah = titik_tengah

    iterasi += 1

tabel_iterasi = pd.DataFrame(
    data_iterasi,
    columns=[
        "Iterasi",
        "Batas Bawah",
        "Batas Atas",
        "Titik Tengah",
        "Nilai Fungsi"
    ]
)

tabel_iterasi
```

## Penjelasan Tabel Iterasi

Setiap baris menunjukkan satu langkah perhitungan Metode Bisection.

Batas bawah dan batas atas merupakan interval pencarian akar.

Titik tengah merupakan nilai yang diperoleh dari rata-rata batas bawah dan batas atas.

Nilai fungsi digunakan untuk menentukan interval baru yang masih mengandung akar.

## Grafik Fungsi

Grafik digunakan untuk membantu melihat bentuk fungsi dan lokasi akar persamaan.

Titik akar yang diperoleh dari Metode Bisection akan ditampilkan pada grafik.

```python
x = np.linspace(1, 3, 400)

y = [hitung_fungsi(nilai_x) for nilai_x in x]

plt.figure(figsize=(8,5))
plt.plot(x, y, label="f(x)", color="green")

plt.axhline(0)

akar = (batas_bawah + batas_atas) / 2

plt.plot(akar, hitung_fungsi(akar), marker="o", color="red")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Grafik Fungsi")
plt.legend()

plt.show()
```

## Interpretasi

Berdasarkan hasil perhitungan menggunakan Metode Bisection, diperoleh pendekatan akar persamaan yang memenuhi tingkat ketelitian sesuai nilai toleransi yang diberikan.

Metode Bisection bekerja dengan membagi interval pencarian menjadi dua bagian secara berulang hingga diperoleh hasil yang mendekati akar sebenarnya.
